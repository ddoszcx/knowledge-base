Задача 3: Иерархия исключений

Условие (Новая, подробная версия):

Вы создаете модуль для работы с внешним API. Ваша задача — спроектировать иерархию исключений для обработки специфических ошибок API и написать функцию-симулятор, которая будет генерировать эти ошибки.

Вам нужно:

Создать базовое исключение ApiError, которое наследует от встроенного Exception.

Создать два дочерних исключения, которые, в свою очередь, наследуют от ApiError:

AuthError(ApiError)

TimeoutError(ApiError)

Написать функцию make_request(should_fail_with: str). Эта функция имитирует запрос к API и должна вести себя следующим образом:

Если should_fail_with равно строке "auth", функция должна сгенерировать (raise) исключение AuthError.

Если should_fail_with равно строке "timeout", функция должна сгенерировать (raise) исключение TimeoutError.

Во всех остальных случаях (если should_fail_with — это любая другая строка или None), функция не должна генерировать исключений (ее тело может просто содержать pass).

```python
class ApiError(Exception):
    pass

class AuthError(ApiError):
    pass

class TimeoutError(ApiError):
    pass 

def make_request(should_fail_with: str):
    if should_fail_with == "auth":
        raise AuthError

    elif should_fail_with == "timeout":
        raise TimeoutError

    else:
        pass 

    
```

Задача 4: Использование raise в методе класса

Условие:
Используйте кастомное исключение для валидации данных внутри метода класса.

Вам нужно:

Создать исключение InvalidAgeError(ValueError). Обратите внимание, что оно наследуется от ValueError — это хорошая практика, когда ваше исключение уточняет стандартное.

Создать класс Person.

В __init__ он должен принимать name и age.

В __init__ нужно проверить возраст: если age меньше 0, необходимо сгенерировать исключение InvalidAgeError с сообщением "Возраст не может быть отрицательным". В противном случае — сохранить атрибуты.

```python
class InvalidAgeError(ValueError):
    pass


class Person:
    def __init__(self, name, age):
        self.name = name
        if age < 0:
            raise InvalidAgeError("Возраст не может быть отрицательным")
        else:
            self.age = age
```

Задача 5: Практический пример "Банкомат"

Условие:
Создайте класс ATM (Банкомат), который использует собственные исключения для управления логикой.

Вам нужно:

Создать базовое исключение AtmError(Exception).

Создать два дочерних исключения: InsufficientFundsError(AtmError) и InvalidPinError(AtmError).

Создать класс ATM.

В __init__ он принимает balance и pin.

Создать метод withdraw(amount, entered_pin). Этот метод должен:

Если entered_pin не совпадает с self.pin, выбросить InvalidPinError.

Если amount больше, чем self.balance, выбросить InsufficientFundsError.

Если все проверки пройдены, уменьшить self.balance на amount и вернуть (return) новый баланс.

```python
class AtmError(Exception):
    pass 

class InsufficientFundsError(AtmError):
    pass

class InvalidPinError(AtmError):
    pass

class ATM:
    def __init__(self, balance, pin):
        self.balance = balance
        self.pin = pin


    def withdraw(self, amount, entered_pin):
        if entered_pin != self.pin:
            raise InvalidPinError

        if amount > self.balance:
            raise InsufficientFundsError

        self.balance = self.balance - amount
        return self.balance
```
